```python
##slicing and indexing
string1 = "Hello "
name = input("Enter your Name: ")
```

```python
string = string1 + name + ", welcome to Python programming"
```

```python
print("a. First character:", string[0])
```

```python
print("b. Last character:", string[-1])
```

```python
print("c. First 5 characters:", string[:5])
```

```python
print("d. Last 11 characters:", string[-11:])
```

```python
print("e. Reverse string:", string[::-1])
```

```python
print("f. Word Python:", string[18:24])
```

```python
print(string.index("Python"))
print(string.index("Python") + 6)
print(string[string.index("Python"):string.index("Python") + 6])
```

```python
print("f. Word Python:", string[29:35])
```

```python
##string methods
strM = "Python beginner tutorial"
```

```python
# Convert to uppercase
print("Uppercase:", strM.upper())
```

```python
# Convert to lowercase
print("Lowercase:", strM.lower())
```

```python
# Capitalize the sentence
print("Capitalize:", strM.capitalize())
```

```python
# Count occurrences of character 't'
print("Count of 't':", strM.count('t'))
```

```python
# Replace Python with Machine Learning
print("Replace:", strM.replace("Python", "Machine Learning"))
```

```python
##tuples creation
tup1 = (10, 20, 30)
tup2 = (40, 50, 60)
```

```python
print(tup1)
```

```python
print(tup2)
```

```python
#concatenate the two tuples
tup_combine = tup1 +tup2
print ("combine_tuple:", tup_combine)
```

```python
#Repeat the elements 3 times
print ("Repeated tuple:", tup_combine * 3)
```

```python
# Access the 3rd element
print("Third element:", tup_combine[2])
```

```python
# Access the first three elements
print("First three elements:", tup_combine[:3])
```

```python
# Access the last three elements
print("Last three elements:", tup_combine[-3:])
```
